```python
# funcion para calcular el xor de dos cadenas de bits
import base64

def xor(a, b): # a y b son cadenas de bits
    return chr(ord(a) ^ ord(b)) # xor de los caracteres de a y b convertidos a enteros
```

```python
# funcion para generar las keystreams
def keystream_generator(key, length): # key es la clave y length es la longitud de la keystream
    keystream = [ord(key[i % len(key)]) for i in range(length)] # generamos la keystream a partir de la clave y la longitud
    return keystream # retornamos la keystream generada
```

```python
# funcion para encriptar un mensaje

def encrypt(message, keystream): # message es el mensaje a encriptar y keystream es la keystream generada
    cifrado = "".join([xor(message[i], chr(keystream[i])) for i in range(len(message))]) # ciframos el mensaje a partir de la keystream
    return base64.b64encode(cifrado.encode()).decode() # retornamos el mensaje cifrado en base64
```

```python
# funcion para desencriptar un mensaje

def decrypt(cifrado, keystream): # cifrado es el mensaje cifrado y keystream es la keystream generada
    cifrado = base64.b64decode(cifrado).decode() # decodificamos el mensaje cifrado en base64
    decifrado = "".join([xor(cifrado[i], chr(keystream[i])) for i in range(len(cifrado))]) # deciframos el mensaje a partir de la keystream generada anteriormente
    return decifrado # retornamos el mensaje decifrado
```

```python
# Probando todas las funciones 

mensaje = "Hola mundo" # mensaje a encriptar
mensaje2 = "holas mundos" # mensaje a encriptar
clave = "hipopotamo" # clave para generar la keystream

keystream = keystream_generator(clave, max(len(mensaje), len(mensaje2))) # generamos la keystream a partir de la clave y la longitud del mensaje
cifrado = encrypt(mensaje, keystream) # ciframos el mensaje
cifrado2 = encrypt(mensaje2, keystream) # ciframos el mensaje
decifrado = decrypt(cifrado, keystream) # deciframos el mensaje
decifrado2 = decrypt(cifrado2, keystream) # deciframos el mensaje

print("------------------------------------------")
print("----------mensaje 1 ---------------------")
print("Mensaje original: ", mensaje)
print("Mensaje cifrado: ", cifrado)
print("Mensaje descifrado: ", decifrado)
print("------------------------------------------")
print("")

print("------------------------------------------")
print("----------mensaje 2 ---------------------")
print("Mensaje original: ", mensaje2)
print("Mensaje cifrado: ", cifrado2)
print("Mensaje descifrado: ", decifrado2)
print("------------------------------------------")

```
