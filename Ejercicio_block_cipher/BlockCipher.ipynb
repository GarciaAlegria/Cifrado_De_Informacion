```python
import math
from PIL import Image
from os import urandom
from Crypto.Cipher import DES, DES3, AES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes
```

# 1. Función de Cifrado y Descifrado DES (ECB)

```python
def encrypt_des(data, key): # algoritmo de cifrado DES
    cipher = DES.new(key, DES.MODE_ECB) # se crea un objeto de cifrado DES con la clave proporcionada en modo ECB
    return cipher.encrypt(pad(data, DES.block_size)) # se cifra el mensaje proporcionado y se añade padding al mensaje
```

```python
def decrypt_des(data, key): # algoritmo de descifrado DES 
    cipher = DES.new(key, DES.MODE_ECB) # se crea un objeto de cifrado DES con la clave proporcionada en modo ECB
    return unpad(cipher.decrypt(data), DES.block_size) # se descifra el mensaje proporcionado y se elimina el padding del mensaje 
```

```python
try:
    # prueba de cifrado y descifrado DES
    with open('des.txt', 'rb') as f:  # se lee el archivo des.txt
        des_data = f.read() # se guarda el contenido del archivo en la variable des_data
    des_key = get_random_bytes(8) # se genera una clave aleatoria de 8 bytes
    encrypted_des = encrypt_des(des_data, des_key) # se cifra el contenido del archivo des.txt
    decrypted_des = decrypt_des(encrypted_des, des_key) # se descifra el contenido cifrado

    print(f"texto original del archivo: {des_data.decode()}")
    print(f"\nDES Key: {des_key.hex()}")
    print(f"\nDES Encrypted: {encrypted_des.hex()}")
    print(f"\nDES Decrypted: {decrypted_des.decode()}")

    with open('encrypted_des.txt', 'wb') as f:
        f.write(encrypted_des)
    with open('decrypted_des.txt', 'wb') as f:
        f.write(decrypted_des)

except Exception as e:
    print(f"Error: {e}")
```

# 2. Función de Cifrado y Descifrado 3DES (CBC)

```python
def encrypt_3des(data, key, iv): # algoritmo de cifrado 3DES
    cipher = DES3.new(key, DES3.MODE_CBC, iv) # se crea un objeto de cifrado 3DES con la clave proporcionada en modo CBC
    padded_data = pad(data, DES3.block_size) # se añade padding al mensaje
    return cipher.encrypt(padded_data) # se cifra el mensaje proporcionado
```

```python
def decrypt_3des(encrypted_data, key, iv): # algoritmo de descifrado 3DES
    cipher = DES3.new(key, DES3.MODE_CBC, iv) # se crea un objeto de cifrado 3DES con la clave proporcionada en modo CBC
    decrypted_data = cipher.decrypt(encrypted_data) # se descifra el mensaje proporcionado
    return unpad(decrypted_data, DES3.block_size) # se elimina el padding del mensaje
```

```python
# Lectura del archivo y proceso de cifrado/descifrado
try:
    with open('3des.txt', 'rb') as f: # se lee el archivo 3des.txt
        des3_data = f.read() # se guarda el contenido del archivo en la variable des3_data
    
    des3_key = DES3.adjust_key_parity(get_random_bytes(24)) # se genera una clave aleatoria de 24 bytes
    des3_iv = get_random_bytes(8) # se genera un vector de inicialización aleatorio de 8 bytes
    
    encrypted_3des = encrypt_3des(des3_data, des3_key, des3_iv) # se cifra el contenido del archivo 3des.txt
    decrypted_3des = decrypt_3des(encrypted_3des, des3_key, des3_iv) # se descifra el contenido cifrado

    # Impresión de resultados
    print(f"Texto original del archivo: {des3_data.decode('utf-8', errors='ignore')}")
    print(f"\n3DES Key: {des3_key.hex()}")
    print(f"\n3DES IV: {des3_iv.hex()}")
    print(f"\n3DES Encrypted (hex): {encrypted_3des.hex()}")
    print(f"\n3DES Decrypted: {decrypted_3des.decode('utf-8', errors='ignore')}")

    # Guardado de resultados
    with open('encrypted_3des.txt', 'wb') as f:
        f.write(encrypted_3des)
    with open('decrypted_3des.txt', 'wb') as f:
        f.write(decrypted_3des)

except Exception as e:
    print(f"Error: {e}")
```

# 3. Generación una función cifrado y descifrado AES con CBC Y ECB

```python
def encrypt_aes_cbc(data, key, iv): # algoritmo de cifrado AES en modo CBC 
    cipher = AES.new(key, AES.MODE_CBC, iv) # se crea un objeto de cifrado AES con la clave proporcionada en modo CBC
    return cipher.encrypt(pad(data, AES.block_size)) # se cifra el mensaje proporcionado y se añade padding al mensaje
```

```python
def decrypt_aes_cbc(data, key, iv): # algoritmo de descifrado AES en modo CBC
    cipher = AES.new(key, AES.MODE_CBC, iv) # se crea un objeto de cifrado AES con la clave proporcionada en modo CBC
    return unpad(cipher.decrypt(data), AES.block_size) # se descifra el mensaje proporcionado y se elimina el padding del mensaje
```

```python
def encrypt_aes_ecb(data, key): # algoritmo de cifrado AES en modo ECB
    cipher = AES.new(key, AES.MODE_ECB) # se crea un objeto de cifrado AES con la clave proporcionada en modo ECB
    return cipher.encrypt(pad(data, AES.block_size)) # se cifra el mensaje proporcionado y se añade padding al mensaje
```

```python
def decrypt_aes_ecb(data, key): # algoritmo de descifrado AES en modo ECB
    cipher = AES.new(key, AES.MODE_ECB) # se crea un objeto de cifrado AES con la clave proporcionada en modo ECB
    return unpad(cipher.decrypt(data), AES.block_size) # se descifra el mensaje proporcionado y se elimina el padding del mensaje
```

```python
try:
    image_path = r"./pic.png" # se define la ruta de la imagen a cifrar
    with open(image_path, 'rb') as image_file: # se abre la imagen en modo lectura binaria
        image_data = image_file.read() # se lee la imagen y se guarda en la variable image_data
    
    aes_key = get_random_bytes(32) # se genera una clave aleatoria de 32 bytes
    aes_iv = get_random_bytes(16) # se genera un vector de inicialización aleatorio de 16 bytes

    encrypted_cbc = encrypt_aes_cbc(image_data, aes_key, aes_iv) # se cifra la imagen en modo CBC 
    decrypted_cbc = decrypt_aes_cbc(encrypted_cbc, aes_key, aes_iv) # se descifra la imagen cifrada en modo CBC
 
    encrypted_ecb = encrypt_aes_ecb(image_data, aes_key) # se cifra la imagen en modo ECB
    decrypted_ecb = decrypt_aes_ecb(encrypted_ecb, aes_key) # se descifra la imagen cifrada en modo ECB

    with open(r'./encrypted_cbc_pic.bin', 'wb') as f:
        f.write(encrypted_cbc)

    with open(r'./decrypted_cbc_pic.png', 'wb') as f:
        f.write(decrypted_cbc)

    with open(r'./encrypted_ecb_pic.bin', 'wb') as f:
        f.write(encrypted_ecb)

    with open(r'./decrypted_ecb_pic.png', 'wb') as f:
        f.write(decrypted_ecb)
    
    print("Imagen cifrada con AES en modo CBC guardada en encrypted_cbc_pic.bin")
    print("Imagen descifrada con AES en modo CBC guardada en decrypted_cbc_pic.png")
    print("Imagen cifrada con AES en modo ECB guardada en encrypted_ecb_pic.bin")
    print("Imagen descifrada con AES en modo ECB guardada en decrypted_ecb_pic.png")

except Exception as e:
    print(f"Error: {e}")
```

# 4. Preguntas a Responder

**¿Qué tamaño de clave se está usando para DES, 3DES y AES?**

R// DES: 8 bytes (64 bits), aunque realmente solo usa 56 bits efectivos para la clave.

3DES: 24 bytes (192 bits), pero internamente usa tres claves de 56 bits cada una.

AES: 32 bytes (256 bits), lo cual es una configuración de clave fuerte para AES.

**¿Qué modo de operación está implementado?**

R// DES: ECB (Electronic Codebook)

3DES: CBC (Cipher Block Chaining)

AES: CBC y ECB

**¿Por qué no debemos usar ECB en datos sensibles?**

R// Porque ECB cifra bloques idénticos de entrada en bloques idénticos de salida, lo que significa que los patrones en los datos originales pueden ser visibles en los datos cifrados.

**¿Cual es la diferencia entre ECB vs CBC, se puede notar directamente en una imagen?**

R// La diferencia principal es que en ECB Cada bloque de texto plano se cifra de manera independiente, mientras que en CBC Cada bloque se cifra usando el resultado cifrado del bloque anterior. Y no, viendo las imagenes no se logra notar directamente en una imagen si fue ECB o CBC.

**¿Que es el IV?**

R// El IV (Vector de Inicialización) es un valor aleatorio que se usa en modos de cifrado como CBC para asegurar que incluso si el mismo texto se cifra con la misma clave, el resultado cifrado será diferente.

**¿Que es el PADDING?**

R// El padding se usa cuando el tamaño de los datos a cifrar no es un múltiplo del tamaño de bloque requerido por el algoritmo (por ejemplo, 8 bytes en DES, 16 en AES). El padding llena el último bloque con bytes adicionales para ajustarlo al tamaño correcto, y luego se elimina después del descifrado.

**¿En qué situaciones se recomienda cada modo de operación?**

R// ECB: No se recomienda para datos sensibles debido a la falta de aleatoriedad. Solo podría usarse en datos sin patrones visibles y donde la rapidez sea más importante que la seguridad.

CBC: Este modo es recomendable para cifrar datos estructurados o imágenes, ya que oculta mejor los patrones y proporciona una seguridad mucho mayor.

Ahora bien otros modos como GCM, CTR: Ofrecen aún más seguridad y características adicionales, como autenticación de datos.

**¿Cómo elegir un modo seguro en cada lenguaje de programación?**

R// Python: Usar librerías como pycryptodome o cryptography, con modos CBC, GCM o CTR, y asegurarse de usar claves y IV generados aleatoriamente como hicimos en este ejercicio.

Java: Usar javax.crypto con modos CBC o GCM, y siempre usar SecureRandom para generar IV y claves.

Go: Las librerías crypto/aes y crypto/cipher son seguras, pero hay que manejar el IV correctamente.

C#: Usar System.Security.Cryptography con AesCryptoServiceProvider y siempre definir IV y claves de forma segura.
