## Ejercicio Block Cipher
### CIFRADO DES

```python
from Crypto.Cipher import DES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
```

```python
def generar_clave_des():
    """Genera una clave DES aleatoria de 8 bytes."""
    return get_random_bytes(8)
```

```python
def cifrar_des_ecb(texto_plano, clave):
    """Cifra un texto plano usando DES en modo ECB."""
    cipher = DES.new(clave, DES.MODE_ECB)
    texto_plano_bytes = texto_plano.encode('utf-8')
    texto_rellenado = pad(texto_plano_bytes, DES.block_size)
    texto_cifrado = cipher.encrypt(texto_rellenado)
    return texto_cifrado
```

```python
def descifrar_des_ecb(texto_cifrado, clave):
    """Descifra un texto cifrado usando DES en modo ECB."""
    cipher = DES.new(clave, DES.MODE_ECB)
    texto_descifrado_rellenado = cipher.decrypt(texto_cifrado)
    texto_descifrado = unpad(texto_descifrado_rellenado, DES.block_size)
    return texto_descifrado.decode('utf-8')
```

```python
# Ejemplo de uso
texto_plano = "Este es un mensaje de prueba para cifrar con DES."
clave = generar_clave_des()

texto_cifrado = cifrar_des_ecb(texto_plano, clave)
print("Texto cifrado:", texto_cifrado)

texto_descifrado = descifrar_des_ecb(texto_cifrado, clave)
print("Texto descifrado:", texto_descifrado)

print("clave:", clave)
```
